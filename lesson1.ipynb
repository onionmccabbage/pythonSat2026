#### Looping, Conditional Logic and Functions

```python
for i in d:
    print(i)

for _ in e: # the colon indicated the start of a code block
    accrued = _*0.02
    fee = _*0.0003
    amount = _ + (accrued-fee)
    print(amount)
# when the indentation ends, that is the end of the code block
```

```python
# Sometimes we need a sequence of numeric values
t = range(-10, 11, 2) # start, stop-before, step=1
for i in t:
    print(i)
```

```python
# conditional logic
t
for i in e:
    if i<0:
        print('negative')
    elif i<3:
        print('less than three')
    elif i>= 0:
        print('positive')
    else:
        print('unknown')
```

```python
# Functions
def fn():
    if d[2]=='aapl':
        return True
    else:
        return False

print ( fn() )

def checkNumeric(n=4): # we may choose to pass a default value
    '''Here we will take a value and check that is it a numeric value
    If it is an int or a float we just return the value
    For anything else we will ...'''
    if type(n)==int or type(n)==float:
        return n
    else:
        return 0 # we may choose to simply return a sensible default value
        # raise TypeError('The value must be a numeric')

# call our function
p = 3.147
result = checkNumeric(p) 
result
```
