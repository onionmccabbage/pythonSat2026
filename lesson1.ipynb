```python
# this is a comment
# we may declare data
a1 = 3    # integer value (int)
b7 = 9.4  # floating point value (float)
thing = '15/09/68' # what is the data type?
welcome = 'hello' # str

type(thing) # we cna check what type of data we are dealing with  
```

```python
# other data types
# collections
# a list is a mutable ordinal collection of any data type
d = ['aapl', 'fb', 'msft', 'nflx']
e = [4, 7.2, 2, 9.9, 2, -4, 9, 235342, 22.8, 75, -2, 5423]
e.append(3)
e.insert(6, 6666666) # also replace(), sort(), reverse(), pop()
print( type(d) ) # we may print anything at any time
d[2] # we use slicing to access a member from a collection
e[3:9] # [start:stop-before]
```

```python
# we may use any data type inside a list
e.extend(d)
# e.pop()
e
```

```python
# tuple is an immutable ordinal collection of any data type
f = ('open','close','high', 'low')
f[0:3]
```

```python
# dictionary: a non-ordinal collection of key:value pairs
g = {'ticker':'sp500', 'start':'20/08/1989', 'end':'25/09/2026' }
g['open'] = 4563
g['close'] = 456435
type(g)
g
```
