```python
# we usually import tools at the top
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sp500 = pd.read_csv('StockData/sp500.csv', parse_dates=True, date_format='%m/%d/%y', index_col='Date')
sp500
```

```python
# Moving Average
sp500['ma_5'] = sp500['Close'].rolling(window=5).mean()
sp500['ma_20'] = sp500['Close'].rolling(window=20).mean()
sp500
```

```python
# resample across ME month end apply some rules
rules = {'Open':'first', 'Close':'last', 'Volume':'sum'}
# here we create a new DataFrame from our existing DataFrame
monthly = sp500.resample('ME').agg(rules)
monthly
```

#### Merge Data from separate sources

```python
aapl = pd.read_csv('StockData/aapl.csv', parse_dates=True, date_format='%m/%d/%y', index_col='Date')
ba = pd.read_csv('StockData/ba.csv', parse_dates=True, date_format='%m/%d/%y', index_col='Date')
```

```python
# Create a new column for the ticker name
aapl['Ticker'] = 'aapl'
aapl
ba['Ticker'] = 'ba'
```

```python
merged = ba.merge(aapl, left_on=['Date'], right_on=['Date'], suffixes=('_ba', '_aapl') )
```
